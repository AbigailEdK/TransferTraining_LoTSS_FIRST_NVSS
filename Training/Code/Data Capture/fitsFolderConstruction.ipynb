## Admin Functions

```python
def check_directory(output_folder, overwrite=True):
    if os.path.exists(output_folder):
        if overwrite:
            shutil.rmtree(output_folder)
            os.makedirs(output_folder)
            return True
        else:
            print(f"Folder {output_folder} already exists. Replace? (y/n)")
            choice = input().lower()

            if choice == 'y':
                shutil.rmtree(output_folder)
                os.makedirs(output_folder)
                return True
            else:
                return False

    else:
        os.makedirs(output_folder)
        return True
    
def count_folder_files(folder, extension=".fits"):
    if extension: 
        number = len([f for f in os.listdir(folder) if f.lower().endswith(extension)])
    else:
        number = len([f for f in os.listdir(folder)])

    print(f"Total FITS files in {folder}: {number}")
    return number
    # return number
```

# Renaming of FITS files & Combining into one folder

## Functions

```python
def rename_fits_files(folder, output_folder, suffix, start=1):
    num_files = count_folder_files(folder)
    # check_directory(output_folder)
    renamed = []
    num_renamed = 0

    for i, filename in enumerate(sorted(os.listdir(folder)), start=start):
        if filename.lower().endswith(".fits"):
            old_path = os.path.join(folder, filename)
            base_name = os.path.splitext(filename)[0]
            new_name = f"{base_name}_{suffix}.fits"
            new_path = os.path.join(output_folder, new_name)

            shutil.copy2(old_path, new_path)
            renamed.append((base_name, filename, new_name, new_path))

            num_renamed += 1
            print(f"Renaming ... {num_renamed}/{num_files}", end="\r")

    print(f"Finished renaming files in {folder}. Total renamed: {len(renamed)}")
```

## Code

```python
print("\n >>> Renaming FIRST files... <<< \n")
rename_fits_files(FIRST_DIR, FITS_DIR, "FIRST", start=1)

print("\n >>> Renaming LOFAR files... <<< \n")
rename_fits_files(LOFAR_DIR, FITS_DIR, "LoTSS", start=1)

print("\n >>> Renaming NVSS files... <<< \n")
rename_fits_files(NVSS_DIR, FITS_DIR, "NVSS", start=1)
```
