```python
from pathlib import Path
import argparse
from astropy.io import fits


def print_fits_dimensions(directory: str) -> None:
    path = Path(directory)

    if not path.exists() or not path.is_dir():
        raise ValueError(f"Invalid directory: {directory}")

    fits_files = sorted(
        [p for p in path.iterdir() if p.is_file() and p.suffix.lower() in {".fits", ".fit", ".fts"}]
    )

    if not fits_files:
        print("No FITS files found.")
        return

    for fits_file in fits_files:
        try:
            with fits.open(fits_file) as hdul:
                # Find first HDU that contains image data
                image_hdu = next((hdu for hdu in hdul if getattr(hdu, "data", None) is not None), None)

                if image_hdu is None:
                    print(f"{fits_file.name}: no image data found")
                    continue

                shape = image_hdu.data.shape  # e.g., (height, width)
                if len(shape) >= 2:
                    height, width = shape[-2], shape[-1]
                    print(f"{fits_file.name}: {width} x {height} pixels")
                else:
                    print(f"{fits_file.name}: shape={shape}")
        except Exception as e:
            print(f"{fits_file.name}: error reading file ({e})")


if __name__ == "__main__":
    print_fits_dimensions("/home/abigaildeklerk/Desktop/DeKlerk_Models/Code/Image Augmentation/RADCAT_F_UNPACKED")
```
